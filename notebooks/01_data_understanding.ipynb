This notebook imports reusable project code from `src/` and uses the real local M5 data when available.


```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks": ROOT = ROOT.parent
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
```

```python
from src.data.loader import load_m5_data
from src.config import settings
data = load_m5_data()
sales, calendar, prices = data["sales"], data["calendar"], data["prices"]
print("sales", sales.shape, "calendar", calendar.shape, "prices", prices.shape)
```

```python
print("Missing values:", {k: int(v.isna().sum().sum()) for k, v in data.items()})
print("Unique products:", sales["item_id"].nunique())
print("Unique stores:", sales["store_id"].nunique())
```

```python
import matplotlib.pyplot as plt
day_cols = [c for c in sales if c.startswith("d_")]
series = sales[day_cols].sum(axis=0)
fig, ax = plt.subplots(figsize=(12,4))
ax.plot(range(len(series)), series.values)
ax.set_title("Aggregate demand over M5 day index")
ax.set_xlabel("Day index")
ax.set_ylabel("Demand")
plt.show()
```

```python
import seaborn as sns
fig, ax = plt.subplots(figsize=(7,4))
sns.histplot(sales[day_cols].to_numpy().ravel(), bins=40, ax=ax)
ax.set_title("Demand distribution")
ax.set_xlabel("Demand")
plt.show()
```
